## Spatial Risk Maps
Generate interactive maps for dashboard integration.

```python
import folium
import json

map_center = [8.5, 39.5]
m = folium.Map(location=map_center, zoom_start=6)
folium.Marker([8.5, 39.5], popup='Oromia Region').add_to(m)
m.save(f'{OUTPUTS_DIR}/maps/risk_map.html')
print('Saved risk map HTML.')
```

## Time Series Plots

```python
import plotly.express as px
import pandas as pd
import numpy as np

dates = pd.date_range('2024-01-01', periods=36, freq='M')
risk = np.random.uniform(0, 1, 36)

fig = px.line(x=dates, y=risk, title='Drought Risk over Time')
fig.write_html(f'{OUTPUTS_DIR}/maps/timeseries.html')
print('Saved timeseries plot.')
```
