## SMS Prototype
Generate early-warning messages from model predictions.

```python
import pandas as pd
import numpy as np

# Mock predictions
locations = ['Zone A', 'Zone B', 'Zone C']
risk_scores = [0.1, 0.6, 0.85]

def get_level(score):
    if score < 0.25: return 'NORMAL'
    if score < 0.5: return 'WATCH'
    if score < 0.75: return 'WARNING'
    return 'SEVERE'

alerts = []
for loc, score in zip(locations, risk_scores):
    level = get_level(score)
    if level in ['WARNING', 'SEVERE']:
        msg = f"DROUGHT {level} — {loc}. Risk score {score:.2f}. Follow local official guidance."
        alerts.append(msg)
        print(msg)
```

## Export Alerts

```python
import json
with open(f'{OUTPUTS_DIR}/reports/alerts.json', 'w') as f:
    json.dump(alerts, f)
print('Alerts exported.')
```
